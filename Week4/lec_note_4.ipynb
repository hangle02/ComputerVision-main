```python
in_ = [1,2,1,3,3,4]
count_ = {}
out_ = in_.copy()
for i in range (len(in_)):
    digit_ = in_[i]
    count_[digit_] = count_.get(digit_,0) + 1

for i in range(len(in_)):
    out_[i] = count_.get(in_[i])


print(out_)
```
